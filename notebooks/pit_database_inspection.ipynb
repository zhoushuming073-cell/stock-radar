# Stock Radar historical database inspection
Read-only companion to the database acceptance report. Aggregate scope is observed common candidate member-days, not a complete US census.

```python
from pathlib import Path
import json, sys
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").is_file() and (p / "data/pit/construction/current.json").is_file())
sys.path.insert(0, str(root / "src"))
from radar.pit.database import HistoricalDatabase
pointer = json.loads((root / "data/pit/construction/current.json").read_text(encoding="utf-8"))
directory = Path(pointer["directory"])
database = HistoricalDatabase(directory)
c = database.connection
database.manifest["source_version"]
```

```python
c.execute("""SELECT membership_status,count(*) AS member_days FROM daily_population WHERE security_type='common' GROUP BY 1 ORDER BY 1""").df()
```

```python
c.execute("""SELECT level,count(*) AS identities FROM identity_quality WHERE security_type='common' GROUP BY 1 ORDER BY 1""").df()
```

```python
scorecard = json.loads((directory / "scorecard.json").read_text(encoding="utf-8"))
{key: scorecard[key] for key in ["price","survivorship_audit","delisted_survival","tier"]}
```
